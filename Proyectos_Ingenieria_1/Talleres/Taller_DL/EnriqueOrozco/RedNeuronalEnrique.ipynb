## Importando librerías

```python
from keras.datasets import mnist
from keras import layers, models
from keras.utils import to_categorical
import numpy as np
import matplotlib.pyplot as plt
```

## Nuestro set de datos

```python
(train_data, train_labels), (test_data, test_labels) = mnist.load_data()
```

## Creando un modelo de datos

```python
model = models.Sequential()
# Agregamos una capa de entrada que aplane la imagen de 28x28 a 784
model.add(layers.Flatten(input_shape=(28, 28, 1)))
model.add(layers.Dense(512, activation='relu'))
model.add(layers.Dense(10, activation='softmax'))
```

```python
model.compile(optimizer='rmsprop',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.summary()
```

## Limpieza de datos

```python
x_train = train_data.reshape((60000,28*28))
x_train = x_train.astype('float32')/255

x_test = test_data.reshape((10000,28*28))
x_test = x_test.astype('float32')/255
```

```python
y_train = to_categorical(train_labels)
y_test = to_categorical(test_labels)
```

## Entrenando el modelo

### Mejora con Data Augmentation
Este proceso crea variaciones aleatorias de las imágenes de entrenamiento para que el modelo sea más flexible.

```python
from keras.src.legacy.preprocessing.image import ImageDataGenerator

# Re-compilar el modelo tras el cambio de arquitectura
model.compile(optimizer='rmsprop',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Preparar los datos en formato 4D para el generador
x_train_img = train_data.reshape(-1, 28, 28, 1).astype('float32') / 255
y_train_cat = to_categorical(train_labels)

# Configurar aumentos
datagen = ImageDataGenerator(
    rotation_range=10,
    zoom_range=0.1,
    width_shift_range=0.1,
    height_shift_range=0.1
)

# Entrenar
print("Entrenando con Data Augmentation...")
model.fit(datagen.flow(x_train_img, y_train_cat, batch_size=128), epochs=5)
```

```python
model.fit(x_train, y_train, epochs=5, batch_size=128)
```

## Evaluando sobre data de test

```python
model.evaluate(x_test,y_test)
```

## Predicción con imágenes externas
Usa la siguiente celda para subir una imagen de un número escrito a mano (preferiblemente trazo oscuro sobre fondo claro o viceversa).

```python
from google.colab import files
from PIL import Image
import io
import numpy as np
import matplotlib.pyplot as plt

# Subir archivo
uploaded = files.upload()

for filename in uploaded.keys():
    # en escala de grises
    img = Image.open(io.BytesIO(uploaded[filename])).convert('L')

    # Redimensionar a 28x28
    img = img.resize((28, 28))

    # Convertir a array de numpy
    img_array = np.array(img)

    # Invertir colores (MNIST es fondo negro, letra blanca)
    img_array = 255 - img_array

    # Preprocesar para el modelo
    img_input = img_array.reshape(1, 28, 28, 1).astype('float32') / 255

    prediction = model.predict(img_input)
    predicted_digit = np.argmax(prediction)
    probabilities = prediction[0] * 100

    # Mostrar resultados visuales
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # 1. Mostrar la imagen procesada
    ax1.imshow(img_array, cmap='gray')
    ax1.set_title(f'Imagen procesada (Predicción: {predicted_digit})')
    ax1.axis('off')

    #probabilidad
    bars = ax2.bar(range(10), probabilities, color='skyblue')
    ax2.set_xticks(range(10))
    ax2.set_ylim(0, 100)
    ax2.set_xlabel('Dígito')
    ax2.set_ylabel('Probabilidad (%)')
    ax2.set_title('Probabilidad por cada número')

    # Resaltar la barra ganadora
    bars[predicted_digit].set_color('orange')

    for i, p in enumerate(probabilities):
        if p > 1:
            ax2.text(i, p + 1, f'{p:.1f}%', ha='center', fontsize=9)

    plt.tight_layout()
    plt.show()

    print(f'El modelo identifica este número como: {predicted_digit} con un {probabilities[predicted_digit]:.2f}% de confianza.')
```
